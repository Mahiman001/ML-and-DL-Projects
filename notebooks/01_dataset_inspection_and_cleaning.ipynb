```python
questionnaire_cols = df.columns[:50]
# 1. Values that occur
print("=== VALUE DISTRIBUTION ===")
print(df[questionnaire_cols].stack().value_counts().sort_index())

# 2. Values outside the documented 1–5 scale
invalid_mask = ~df[questionnaire_cols].isin([1, 2, 3, 4, 5])

print("\n=== INVALID VALUES ===")
print("Total invalid values:", invalid_mask.sum().sum())

# 3. Invalid values by question
print("\n=== INVALID VALUES BY QUESTION ===")
print(invalid_mask.sum().sort_values(ascending=False))

# 4. Missing values
print("\n=== MISSING VALUES ===")
print("Total missing:", df[questionnaire_cols].isna().sum().sum())
```

```python
# Reverse-scored questionnaire items
reverse_items = [
    "EXT2", "EXT4", "EXT6", "EXT8", "EXT10",
    "AGR1", "AGR3", "AGR5", "AGR7", "AGR9",
    "CSN2", "CSN4", "CSN6", "CSN8", "CSN10",
    "EST1", "EST3", "EST5", "EST7", "EST9",
    "OPN2", "OPN4", "OPN6", "OPN8", "OPN10"
]

# Trait → questionnaire items
trait_items = {
    "E": [f"EXT{i}" for i in range(1, 11)],
    "A": [f"AGR{i}" for i in range(1, 11)],
    "C": [f"CSN{i}" for i in range(1, 11)],
    "ES": [f"EST{i}" for i in range(1, 11)],
    "O": [f"OPN{i}" for i in range(1, 11)]
}
```

THIS IS VERY IMPORTANT HERE WE SEE THAT EACH TRAIT HAS 10 QUESTIONS BUT THE YES ANSWER OR 5POINTS 
IS INVERSLY RELATED TO THE QUESTION so we  have to reverse the point 

Question
   ↓
Does agreeing with it mean MORE of the trait?
   ↓
YES → keep the score
NO  → reverse the score
   ↓
Combine the 10 questions
   ↓
One trait score

```python
# Copy questionnaire data
scores = df[questionnaire_cols].copy()
# Reverse-score the reverse-keyed questions
for col in reverse_items:
    scores[col] = 6 - scores[col]

scores.head()
```

```python
# Calculate trait scores
trait_scores = pd.DataFrame(index=scores.index)

trait_scores["E"] = scores[trait_items["E"]].mean(axis=1)
trait_scores["A"] = scores[trait_items["A"]].mean(axis=1)
trait_scores["C"] = scores[trait_items["C"]].mean(axis=1)
trait_scores["ES"] = scores[trait_items["ES"]].mean(axis=1)
trait_scores["O"] = scores[trait_items["O"]].mean(axis=1)

trait_scores.head()
```

```python
trait_valid_counts = pd.DataFrame(index=scores.index)
for trait, cols in trait_items.items():
    trait_valid_counts[trait] = scores[cols].notna().sum(axis=1)
trait_valid_counts.min(axis=1).value_counts().sort_index()
```

 WE HAVE 10 QUESTIONS THIS IS SHOWING PEOPLE WITH Minimum valid answers across 5 traits
 1696 people have 0 valid answers 

 SO THE RULE IS Require at least 8 valid answers out of 10 for every trait.

```python
valid_for_all_traits = trait_valid_counts.min(axis=1) >= 8
print("Participants before:", len(df))
print("Participants kept:", valid_for_all_traits.sum())
print("Participants removed:", (~valid_for_all_traits).sum())
```

```python
trait_scores_clean = trait_scores.loc[valid_for_all_traits].copy()
trait_scores_clean.shape
```

```python
trait_scores_clean.describe()
```

ES = Emotional Stability
N  = Neuroticism
WE HAVE TO CONVERT THE ABOVE THINGS 

ONE MORE THING    High ES → Low N
Low ES  → High N

```python
trait_scores_clean["N"] = 6 - trait_scores_clean["ES"]

trait_scores_clean = trait_scores_clean.drop(columns="ES")

trait_scores_clean.head()
```

```python
# Create final dataset for ML training
model_data = pd.concat(
    [
        scores.loc[trait_scores_clean.index, questionnaire_cols],
        trait_scores_clean[["E", "A", "C", "O", "N"]]
    ],
    axis=1
)

# Save processed dataset
model_data.to_csv(
    "../DATA/processed/personality_model_data.csv",
    index=False
)

print("Saved successfully!")
print("Shape:", model_data.shape)
```
